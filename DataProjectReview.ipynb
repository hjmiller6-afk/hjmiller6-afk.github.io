# **Data Project Review** 

By: Hart Miller

*In Pursuit of Democracy* is a data project published by Alvin Chang in 2025. In this project, Chang is attempting to tell a story that grasps every time the word “democracy” was said in the Congressional Record since 1880, a record compiled of verbatim transcripts of the proceedings, debates, and actions of the United States Congress. Using specific examples and famous moments in our country’s history of pursuing democracy, Chang highlights how the meaning and emphasis of democracy has evolved over time. While he keeps record of every time the word appears, he also distinguishes when the word has been used to argue that democracy is under threat during specific situations or periods of time (suffrage movements, war, voter fraud, etc). 

**Analysis & Interpretation**


The point of view largely revolves around the argument which utilizes the data that constitutes democracy being used in terms of it being under threat. With this, Chang makes the point that “democracy under threat” is a recurring theme in American politics and speech, and that the nation’s identity lies in the continued, contested pursuit of democratic ideals. By highlighting specific moments in history, it becomes more clear how America has continued to strive for an evolving version of democracy. Periods of time where voting rights were argued not only for but also against, moments when war made Americans fear for the safety of their government, and even declines in belief in the integrity of elections all portray how threats to democracy have always been a fear of our country. 

The argument is clear and effectively communicated. From the visual above, it is obvious what the goal of the creator is, which again is how consistent feelings towards threats to our democracy have been since the late 19th century. Furthermore, the final message from the creator ties the story together by explaining how democracy today is just as important as it was when the founding fathers wrote that every man is equal. 


**Data & Evidence**

Chang explains using a massive amount of data to make his argument. In his section on “Data and Methods”, For every year since 1880, speeches with the word "democracy" were analyzed with a Google LLM model, and for every speech it analyzed the 200 words before and after the prompt in order to distinguish whether or not the speaker felt democracy was under threat. Chang uses a prompt with this data to ensure that speeches are not counted twice. This data is used very effectively in my opinion, as the data presents both how often the word is used as well as when it is used under the circumstances which the prompt outlines. Also, by using the data alongside the speeches and context of the time periods, the quantitative data is supported effectively. 

**Organization & Design**

I found the project to be well organised, but I did find an issue with how accessible it was. As the purpose of the data was easily interpreted, credit should be given to the organisation of it. It was easy to navigate with a continuous scrolling method, as well as clear guidelines of exactly how the data should be interpreted. Chang does an excellent job of pointing out importances in the data, and also pointing out different tools used to exploit important parts of the story. More importantly, these tools were pointed out when they could be most effectively used to make the purpose more obvious. The main critique I have falls on the area of accessibility and the method that Chang uses to observe the data. While the scrolling trait did help organize the observation of data chronologically, allowing the feeling of a story, it also made it more difficult to jump around immediately. This was an issue for me because it made comparing the use of the word democracy in a period such as the late 19th century with its use in the 21st a more tedious task than I would have hoped for, because I would need to scroll through countless speeches to answer my curiosity every time I had a question.


**Communication & Audience**


Compiling my review on the interpretation, use of data and evidence, and organization of this data project, I believe that the project’s argument and evidence is evident and communicated well. Although Chang’s message isn’t obvious right away, I had no trouble understanding what was trying to be conveyed about democracy by the end, especially with his remarks given at the conclusion. I do believe there is a clear audience. With occasional declining faith in our democracy, younger generations can feel that the country is moving in the wrong direction, but in reality this data shows that preserving our country’s principles has always been of the utmost importance, and younger Americans should be reminded that striving to make our country better does not make us a poorly ran country.
